Faça um algoritmo para ler dois números

### algoritmo da soma

1. Leia o primeiro número e guarde numa variável chamada numero01
2. Leia o segundo número e guarde em outra variável chamada numero02

3. faça a soma das duas variavel e guarde o resultado chamada soma

4. mostre o resultado

```python
Numero01 = int(input('Digite o primeiro número: '))
Numero02 = int(input('Digite o segundo número: '))
Soma = Numero01 + Numero02
print(f'O resultado é: {Soma}')
```

### algoritmo da média

2 faça um algoritmo para ler três números e imprimir a soma, média e produtos dos números lidos.

1. Leia o primeiro número e salve na variável num1
2. Leia o segundo número e salve na variável num2
3. Leia o terceiro número e salve na variável num3
4. Some os 3 números e salve na variável soma
5. Mostre o resultado da soma

```python
num1 = int(input("digite o primeiro número:"))
num2 = int(input("digite o segundo número:"))
num3 = int(input("digite o terceiro número:"))
soma = num1 + num2 + num3
print(f"a soma dos números é {soma}")
```

### algoritmo do maior e menor

1. Leia o primeiro número e salve na variável num1
2. Leia o segundo número e salve na variável num2
3. Descobrir o maior
4. Mostre o menor
5. Descobrir o menor
6. Mostrar o menor
7. Descobrir se são iguais
8. Mostrar se são iguais

```python
num1 = int(input("digite o primeiro número:"))
num2 = int(input("digite o segundo número:"))
if num1 > num2:
    print(f"o maior número é {num1}")
    print(f"o menor número é {num2}")
elif num2 > num1:
    print(f"o maior número é {num2}")
    print(f"o menor número é {num1}")
else:
    print("os números são iguais")
```
